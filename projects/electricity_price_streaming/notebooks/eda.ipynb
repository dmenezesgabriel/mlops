# Exploratory analysis: a market that drifts

45,312 half-hour ticks, seven market features, one binary direction label.
How do price and demand behave — and where does the relationship shift?

```python
import matplotlib.pyplot as plt
from electricity_price_streaming.prepare_data import load_frame

frame = load_frame()
frame.describe().round(3)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 3.5))
frame["nswprice"].iloc[:4000].plot(
    ax=axes[0], title="nswprice (first 4000 ticks)"
)
frame["class"].value_counts().plot.bar(ax=axes[1], title="direction balance")
plt.tight_layout()
```

```python
frame.groupby("period")["nswprice"].mean().plot(
    figsize=(7, 3), title="price by half-hour period"
)
```

**Drift is the story**: the market rules changed mid-1997 and the
price/demand relationship visibly shifts — the reason this dataset is a
streaming benchmark. The model trained on early ticks faces genuinely
different later ticks, which is what a real feed looks like.

```python
frame[["nswprice", "nswdemand"]].rolling(500).mean().plot(
    figsize=(9, 3), title="rolling means — watch the regime shift"
)
```
