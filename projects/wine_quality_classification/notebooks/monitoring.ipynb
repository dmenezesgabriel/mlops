# Monitoring + teardown

Pipeline health is step success rate and metric trend across runs; locally,
list the executions and the artifacts, then tear the stack down.

```python
from sagemaker.local.local_session import LocalPipelineSession

session = LocalPipelineSession()
client = session.boto_session.client("sagemaker")
[
    p["PipelineExecutionArn"]
    for p in client.list_pipeline_executions(
        PipelineName="wine-cultivar-pipeline"
    ).get("PipelineExecutionSummaries", [])
]
```

```python
# inspect the latest run's steps
summaries = client.list_pipeline_execution_steps(
    PipelineExecutionArn=client.list_pipeline_executions(
        PipelineName="wine-cultivar-pipeline"
    )["PipelineExecutionSummaries"][0]["PipelineExecutionArn"]
)["PipelineExecutionSteps"]
[(s["StepName"], s["StepStatus"]) for s in summaries]
```
