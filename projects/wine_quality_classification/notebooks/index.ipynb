# Wine quality classification — the project at a glance

Cultivar classification from lab assays, built as a **SageMaker Pipeline**
(with `LocalPipelineSession`) fed by a **Feast offline store** — the
deployment shape for routine, auditable retraining.

Same production lifecycle as the NYC taxi project: frame → collect →
preprocess → explore → store → train → tune → evaluate → deploy → monitor.

## The lifecycle

| Stage | Notebook | What it does |
|---|---|---|
| Frame | `docs/01_problem_framing.md` | Why a managed pipeline |
| Collect | `docs/02_data_collection.md` | The wine dataset |
| Preprocess | `docs/03_data_preprocessing.md` | Entity keys + scaler-in-model |
| Explore | `eda.ipynb` | What separates the cultivars |
| Feature store | `feature_engineering.ipynb` | Feast apply + retrieval |
| Train | `training.ipynb` | Single job, as a preview of the step |
| Tune | `tuning.ipynb` | `C` sweep as repeated jobs |
| Evaluate | `evaluation.ipynb` | Holdout accuracy gate |
| Deploy | `deploy.ipynb` | **The pipeline**: train → eval → condition → register |
| Monitor | `monitoring.ipynb` | Run history + teardown |

## Architecture

```
prepare_data → feature_repo (Feast offline)
                    │ get_historical_features → S3 channel
                    ▼
   Pipeline: TrainingStep → evaluation metrics → ConditionStep → register
```

The pipeline is the product: every retrain is one `pipeline.start()`, and
each step is a logged, inspectable job on moto.
