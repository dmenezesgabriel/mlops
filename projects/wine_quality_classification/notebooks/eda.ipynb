# Exploratory analysis: what separates the cultivars

178 samples, 13 chemistry measures, 3 classes. How separable are they?

```python
import matplotlib.pyplot as plt
from wine_quality_classification.prepare_data import load_frame

frame = load_frame()
frame.describe().round(2)
```

```python
frame["cultivar"].value_counts().sort_index().plot.bar(
    figsize=(4, 3), title="class balance"
)
```

```python
frame.corr(numeric_only=True)["cultivar"].drop(
    "cultivar"
).sort_values().plot.barh(figsize=(7, 4), title="correlation with cultivar")
```

`flavanoids` and `od280_od315` carry the most signal — phenolic
chemistry, exactly what an enologist would name first. A linear model on
scaled features should separate the three cultivars almost completely.

```python
fig, ax = plt.subplots(figsize=(6, 4))
for c, grp in frame.groupby("cultivar"):
    ax.scatter(
        grp["flavanoids"], grp["color_intensity"], s=14, label=f"cultivar {c}"
    )
ax.legend()
ax.set_xlabel("flavanoids")
ax.set_ylabel("color_intensity")
```
