# Hyperplane visualization

A hyperplane in $\mathbb{R}^n$ is the set of points satisfying $\mathbf{w}^T \mathbf{x} + b = 0$.
In 2D it's a line; in 3D it's a plane. Here we plot both.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
w = np.array([2.0, 1.0])
b = -3.0

xs = np.linspace(-5, 5, 100)
ys = (-w[0] * xs - b) / w[1]

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(xs, ys, color="crimson", lw=2, label=f"{w[0]}x + {w[1]}y + {b} = 0")
ax.fill_between(xs, ys, ys + 2, color="tab:blue", alpha=0.15)
ax.fill_between(xs, ys, ys - 2, color="tab:orange", alpha=0.15)
ax.axhline(0, color="gray", lw=0.8)
ax.axvline(0, color="gray", lw=0.8)
ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
ax.set_aspect("equal")
ax.set_title("Hyperplane in R^2 (a line)")
ax.legend()
plt.show()
```

```python
w3 = np.array([1.0, 1.0, 1.0])
b3 = -2.0

xx, yy = np.meshgrid(np.linspace(-5, 5, 40), np.linspace(-5, 5, 40))
zz = (-w3[0] * xx - w3[1] * yy - b3) / w3[2]

fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(xx, yy, zz, alpha=0.6, cmap="coolwarm")
ax.set_title("Hyperplane in R^3 (a plane)")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()
```
